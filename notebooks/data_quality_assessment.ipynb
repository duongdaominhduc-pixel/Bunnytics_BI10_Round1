# Data Quality Assessment
Phân tích chất lượng dữ liệu với 6 tasks: Missing, Duplicates, Outliers, Consistency, Temporal, Cross-dataset.
*(Theo concept màu Cam - Orange)*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys

# Thêm src vào path để gọi visualization
sys.path.append(os.path.abspath(os.path.join('..')))
from src.visualization import COLORS, set_style
set_style()

DATA_DIR = os.path.join('..', 'ĐỀ BÀI', 'DATASET')

print('Đang load dữ liệu...')
df_health = pd.read_csv(os.path.join(DATA_DIR, 'consumer_financial_health_engagement_2025.csv'))
df_txn = pd.read_csv(os.path.join(DATA_DIR, 'consumer_transactions_2025.csv'))
print(f'Load thành công! Health: {df_health.shape[0]:,} dòng | Txn: {df_txn.shape[0]:,} dòng')
```

### 1 & 2: Missing Values & Duplicates
Dữ liệu không bị missing (ngoại trừ `next_month_low_health_flag`) và hoàn toàn không có duplicates.

### 3. Outliers (spend_amount_vnd)
Kiểm tra biểu đồ Box Plot để xem sự phân bố của số tiền chi tiêu.

```python
plt.figure(figsize=(10, 4))
sns.boxplot(x=df_txn['spend_amount_vnd'], color=COLORS['primary'])
plt.title('Phân bố số tiền giao dịch (Spend Amount VND)')
plt.xlabel('Số tiền (VND)')
plt.show()
```

### 4. Temporal Coverage (Phân bố giao dịch theo tháng)
Xem xét số lượng giao dịch trong 12 tháng của năm 2025.

```python
df_txn['activity_datetime'] = pd.to_datetime(df_txn['activity_datetime'])
df_txn['month'] = df_txn['activity_datetime'].dt.month

monthly_counts = df_txn['month'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
sns.barplot(x=monthly_counts.index, y=monthly_counts.values, color=COLORS['accent1'])
plt.title('Số lượng giao dịch theo tháng trong năm 2025')
plt.xlabel('Tháng')
plt.ylabel('Số lượng giao dịch')
plt.show()
```
